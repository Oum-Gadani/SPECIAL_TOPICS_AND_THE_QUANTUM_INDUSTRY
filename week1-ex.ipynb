## Excercise 1: Energy Levels of Hydrogen

```python
h = 6.62607015e-34
c = 299792458
pi = 3.14159265359
hbar = h/(2 * pi)
e = 1.602176634e-19
me = 9.10938370e-31
ep0 = 8.85418781e-12
```

Use these to calculate the Bohr radius,

$$a_0 = \frac{4\pi \epsilon_{0} \hbar^2}{m_e e^2} = \frac{\hbar}{m_e c \alpha} $$

```python
a0 = 4 * pi * ep0 * hbar**2 / (me * e**2)
print("Bohr radius (a0):", a0, "m")
```

The energy of the $n$-th is given by

$$ E_n = - \frac{e^2}{8 \pi \varepsilon_0 a_0 n^2} = \frac{E_1}{n^2} $$
Make a variable for the $n=1$ energy level.

```python
def energy(n):
    return -(e**2/(8*pi*ep0*a0*n**2))


n=1

print("Energy of Hydrogen Atom at n=1:", energy(n))
```

Change to electron volts (eV) by deviding by $e$.

```python
def change_to_eV(energy):
    return energy / e
```

Make a list with this as the only element to store the energy levels.

```python
energy_lvl_list = []
print("Data Type of energy_lvl_list: " + str(type(energy_lvl_list)))
```

Append this list with the next 5 values for $n$.

```python
for i in range(2, 7):
    energy_lvl_list.append(i)

print("Energy levels in the list:", energy_lvl_list)
```

Cast this list into a set to remove any duplicates you may have created.

```python
energy_lvl_list = set(energy_lvl_list)
print("Data Type of energy_lvl_list after conversion to set: " + str(type(energy_lvl_list)))
print("Energy levels in the set:", energy_lvl_list)
```

Now cast back into a list.



```python
energy_lvl_list = list(energy_lvl_list)
print("Data Type of energy_lvl_list after conversion to list: " + str(type(energy_lvl_list)))
```

Sort the list.

```python

energy_lvl_list.sort()
print("Energy levels in the sorted list:", energy_lvl_list)
```

This is however in the reverse order to what we want so reverse it.

```python
energy_lvl_list.sort(reverse=True)
print("Energy levels in the sorted list in descending order:", energy_lvl_list)
```

Now calculate the first two wavelengths (i=3,4 f=2) of photons emitted in the visible Balmer series of transitions.

$$ \lambda = hc/E_\gamma$$

where

$$E_\gamma = E_i - E_f $$

the final energy state for the Balmer series is 

$$ f=2 $$

```python
E2 = energy(2)
E3 = energy(3)
E4 = energy(4)

E_gamma_3_2 = E3 - E2
E_gamma_4_2 = E4 - E3

lambda_3_2 = h * c / E_gamma_3_2
lambda_4_2 = h * c / E_gamma_4_2

print("Wavelength (3->2):", lambda_3_2)
print("Wavelength (4->2):", lambda_4_2)
```

We are in eV, to convert back to S.I, divide the wavelengths by $e$ and print.

```python
SI_3_2 = change_to_eV(E_gamma_3_2)
SI_4_2 = change_to_eV(E_gamma_4_2)

print("Energy (3->2):", SI_3_2)
print("Energy (4->2):", SI_4_2)
```

It turns out we can contain lists within lists i.e. [4.0,[1,5]]. Using nested lists for each row, represent the following single quibit gates X,Y, and Z and print them.

$$ X = \begin{pmatrix}
0 & 1 \\ 1&0

Y = \begin{pmatrix}

Z = \begin{pmatrix}

```python
pauli_x = [[0, 1],
           [1, 0]]

pauli_y = [[0, -1j],
           [1j, 0]]

pauli_z = [[1, 0],
           [0, -1]]

```

Now, extract and print the element from the first row and second column of the Y gate.

Now attempt to multiply two of the matrices

```python
mul_x_y = [[0, 0],
            [0, 0]]

for i in range(2):
    for j in range(2):
        for k in range(2):
            mul_x_y[i][j] += pauli_x[i][k] * pauli_y[k][j]

print("Multiplication of Pauli X and Y matrices:")
for i in range(2):
    for j in range(2):
        print(mul_x_y[i][j], end="\t")
    print()
```
